```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='City', hue='new ratings', palette='Set1')
plt.title('Restaurant Count by Rating Tier per City')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.legend(title='Rating Tier')
plt.show()
```

```python
# Calculate counts per city
city_counts = df['City'].value_counts()

# Plot Pie Chart
plt.figure(figsize=(8, 8))
plt.pie(city_counts, labels=city_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Percentage Share of Restaurants by City')
plt.ylabel('') # Hides the default y-axis label
plt.show()
```
